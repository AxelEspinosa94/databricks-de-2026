# Exercise 5: Use of OPTIMIZE and ZORDER functions

To perform this exercise we will make use of the delta tables built in exercises 1, 2 and 4, all of them are a set of parquet files by definition.

We will begin with the table defined in exercise 1 `workspace.default.streaming_demo`

```python
%sql
-- First we have to convifmr that the table exists in the Unity Catalog and it is managed
DESCRIBE DETAIL workspace.default.streaming_demo;
```

```python
%sql
 -- Then we execute OPTIMIZE + ZORDER
OPTIMIZE workspace.default.streaming_demo
 ZORDER BY (id);
```

```python
%sql
DESCRIBE DETAIL workspace.default.streaming_demo;
```

The number of files decreased so OPTIMIZE reduced the parquet files of the table.

Now let's do the same for the delta table in exercise 2 `workspace.default.streaming_demo_transformed`

```python
%sql
OPTIMIZE workspace.default.streaming_demo_transformed
ZORDER BY (id);
```

```python
%sql
DESCRIBE DETAIL workspace.default.streaming_demo_transformed;
```
